# 1. Business Understanding

## 1.1 Latar Belakang
Hipertensi atau tekanan darah tinggi adalah salah satu faktor risiko utama untuk berbagai penyakit kardiovaskular. Salah satu faktor gaya hidup yang paling sering dikaitkan dengan hipertensi adalah tingkat konsumsi garam (natrium). 

Penelitian ini bertujuan untuk melakukan analisis mendalam mengenai pengaruh tingkat konsumsi garam dan faktor-faktor lain (seperti usia, tingkat stres, riwayat hipertensi, BMI, dll) terhadap risiko terjadinya hipertensi.

## 1.2 Tujuan Proyek
1. Memahami karakteristik data dan distribusi setiap fitur.
2. Mengidentifikasi hubungan antara tingkat konsumsi garam dengan kejadian hipertensi.
3. Membangun model klasifikasi menggunakan algoritma **LightGBM** untuk memprediksi risiko hipertensi.
4. (Fase Selanjutnya) Menginterpretasikan model menggunakan **SHAP** untuk memahami kontribusi setiap fitur terhadap prediksi model.

## 1.3 Deskripsi Fitur
- `Age`: Usia pasien (Tahun).
- `Salt_Intake`: Estimasi asupan garam harian.
- `Stress_Score`: Skor tingkat stres pasien (0-10).
- `BP_History`: Riwayat tekanan darah (Normal, Prehypertension, Hypertension).
- `Sleep_Duration`: Durasi tidur harian (Jam).
- `BMI`: Body Mass Index.
- `Medication`: Jenis obat yang sedang dikonsumsi.
- `Family_History`: Riwayat hipertensi di keluarga (Yes/No).
- `Exercise_Level`: Tingkat aktivitas fisik (Low, Moderate, High).
- `Smoking_Status`: Status merokok (Smoker, Non-Smoker).
- `Has_Hypertension`: Variabel Target (Yes/No).
